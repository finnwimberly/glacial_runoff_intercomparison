```python
mport xarray as xr
import glob

# template for reading batched pygem sims

 # get list of files containing annual glacier area for region 15
ds_fns = glob.glob('/path/to/compiled/simulations/R15_glac_area_annual_*.nc')
ds_fns = '/Users/finnwimberly/Library/CloudStorage/GoogleDrive-fwimberly@middlebury.edu/My Drive/Lizz Research Stuff/Precipitation Analysis/'

# check for batched outputs so we can aggregate them
ds_fns_int = []                
for i in ds_fns:
    if 'Batch' in i:
        ds_fns_int.append(int(i.split('-')[-2]))

# sort filenames
if len(ds_fns_int) > 0:
    ds_fns = [x for _,x in sorted(zip(ds_fns_int, ds_fns))]

# load data (aggregate if necessary)
for nfn, ds_fn in enumerate(ds_fns):
    if nfn == 0:
        ds = xr.open_dataset(ds_fn)
    else:
        ds_batch = xr.open_dataset(ds_fn)
        ds = xr.concat((ds, ds_batch), dim='glacier')
```
